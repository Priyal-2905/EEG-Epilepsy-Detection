# EEG Recording Visualizer

```python
import numpy as np
import matplotlib.pyplot as plt

FILE_PATH   = "./chb/Seizure/ictal/1.csv"
N_POINTS    = 1280
SAMPLE_RATE = 256   


data = np.loadtxt(FILE_PATH, delimiter=",", max_rows=N_POINTS)
n_samples, n_channels = data.shape
time = np.arange(n_samples) / SAMPLE_RATE

print(f"File    : {FILE_PATH}")
print(f"Shape   : {data.shape}  ({n_channels} channels × {n_samples} samples)")
print(f"Duration: {n_samples / SAMPLE_RATE:.3f} s")
```

```python

data_norm = (data - data.mean(axis=0)) / (data.std(axis=0) + 1e-8)

ch_labels = [f"CH-{i+1:02d}" for i in range(n_channels)]
spacing   = 4.0
offsets   = np.arange(n_channels) * spacing

fig, ax = plt.subplots(figsize=(16, n_channels * 0.5 + 2))

for i in range(n_channels):
    ax.plot(time, data_norm[:, i] + offsets[i],
            color="#2563eb", linewidth=0.7)

ax.set_yticks(offsets)
ax.set_yticklabels(ch_labels, fontsize=8)
ax.set_xlabel("Time (s)", fontsize=10)
ax.set_title("EEG Recording — Seizure / Ictal (file 1)\n"
             f"{n_channels} channels · {N_POINTS} samples · "
             f"{N_POINTS/SAMPLE_RATE:.1f} s @ {SAMPLE_RATE} Hz",
             fontsize=11)

ax.grid(axis="x", color="#e5e7eb", linewidth=0.5)
ax.set_axisbelow(True)
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

plt.tight_layout()
plt.savefig("eeg_plot.png", dpi=150, bbox_inches="tight")
plt.show()
```
